```python
def quadratic_approx(f, x1, dx, eps=1e-5, max_iter=1000):
    i = 0

    while i < max_iter:
        print(f"\n === Итерация {i} ===")
        x2 = x1 + dx
        x3 = x1 - dx
        print(f"x1:{x1:.3f}   x2:{x2:.3f}   x3:{x3:.3f}")

        f1, f2, f3 = f(x1), f(x2), f(x3)
        print(f"f1:{f1:.3f}   f2:{f2:.3f}   f3:{f3:.3f}")

        # выбираем лучшую точку
        if f1 <= f2 and f1 <= f3:
            x_min, f_min = x1, f1
        elif f2 <= f1 and f2 <= f3:
            x_min, f_min = x2, f2
        else:
            x_min, f_min = x3, f3
        print(f"F_min: {f_min:.3f}   x_min: {x_min:.3f}")

        # знаменатель параболы
        denom = (x2 - x3) * f1 + (x3 - x1) * f2 + (x1 - x2) * f3

        if denom == 0:
            x1 = x_min
            i += 1
            continue

        # вершина параболы
        x_m = 0.5 * (
            (x2**2 - x3**2) * f1 +
            (x3**2 - x1**2) * f2 +
            (x1**2 - x2**2) * f3
        ) / denom

        f_m = f(x_m)

        print(f"F_m: {x_m:.3f}   x_m: {f_m:.3f}")

        # критерий остановки
        print(abs(x_m - x_min), abs(f_m - f_min))
        if abs(x_m - x_min) < eps and abs(f_m - f_min) < eps:
            return x_m, i

        # обновляем базовую точку
        x1 = x_m
        i += 1
        print(x1)

    return x1, i
```

```python
f = lambda x: x**2 - 3*x + x*math.log(x, math.e)

quadratic_approx(f, 1.5, 0.2, 0.01)
```
