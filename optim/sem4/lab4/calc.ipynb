```python
import numpy as np

# Функция
def f(x):
    x1, x2 = x
    return 3*x1**2 + 2*x1*x2 + 4*x2**2 + 12*x1 + 26*x2 + 45

# Градиент функции
def grad_f(x):
    x1, x2 = x
    df_dx1 = 6*x1 + 2*x2 + 12
    df_dx2 = 2*x1 + 8*x2 + 26
    return np.array([df_dx1, df_dx2])
```

```python
def coordinate_descent(x0, eps=1e-6, max_iter=100):
    x = np.array(x0, dtype=float)
    n = len(x)
    
    for _ in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            # Минимизация по одной координате
            def f_i(xi):
                x_temp = x.copy()
                x_temp[i] = xi
                return f(x_temp)
            # Производная по xi
            df_i = lambda xi: grad_f(np.array([x[0], x[1]]))[i]
            
            # Шаг градиентного спуска по одной координате
            alpha = 0.01
            xi = x[i]
            xi_new = xi - alpha * df_i(xi)
            x[i] = xi_new
        
        if np.linalg.norm(x - x_old) < eps:
            break
    return x

x_cd = coordinate_descent([2,4])
print("Coordinate Descent:", x_cd)
```

```python
def gradient_descent(x0, alpha=0.05, tol=1e-6, max_iter=100):
    x = np.array(x0, dtype=float)
    for _ in range(max_iter):
        g = grad_f(x)
        x_new = x - alpha * g
        if np.linalg.norm(x_new - x) < tol:
            break
        x = x_new
    return x

x_gd = gradient_descent([2,4])
print("Gradient Descent:", x_gd)
```

```python
A = np.array([[6,2],[2,8]])

def steepest_descent(x0, tol=1e-6, max_iter=100):
    x = np.array(x0, dtype=float)
    for _ in range(max_iter):
        g = grad_f(x)
        alpha = (g @ g) / (g @ (A @ g))
        x_new = x - alpha * g
        if np.linalg.norm(x_new - x) < tol:
            break
        x = x_new
    return x

x_sd = steepest_descent([2,4])
print("Steepest Descent:", x_sd)
```

```python
import numpy as np

# Функция и градиент
def f(x):
    x1, x2 = x
    return 3*x1**2 + 2*x1*x2 + 4*x2**2 + 12*x1 + 26*x2 + 45

def grad_f(x):
    x1, x2 = x
    return np.array([6*x1 + 2*x2 + 12, 2*x1 + 8*x2 + 26])

# -------------------------------
# Метод покоординатного спуска
# -------------------------------
def coordinate_descent_steps(x0, steps=3):
    x = np.array(x0, dtype=float)
    print("Метод покоординатного спуска:")
    for k in range(steps):
        # Минимизация по x1
        x1_old = x[0]
        x[0] = (-2*x[1] - 12)/6  # df/dx1 = 0
        print(f"Шаг {k+1}.1: x1: {x[0]:.6f}, x2: {x[1]:.6f}")
        # Минимизация по x2
        x[1] = (-2*x[0] - 26)/8  # df/dx2 = 0
        print(f"Шаг {k+1}.2: x1: {x[0]:.6f}, x2: {x[1]:.6f}")
    return x

# -------------------------------
# Метод градиентного спуска
# -------------------------------
def gradient_descent_steps(x0, alpha=0.1, steps=3):
    x = np.array(x0, dtype=float)
    print("\nМетод градиентного спуска:")
    for k in range(steps):
        g = grad_f(x)
        x = x - alpha * g
        print(f"Шаг {k+1}: x: ({x[0]:.6f}, {x[1]:.6f}), градиент: ({g[0]:.6f}, {g[1]:.6f})")
    return x

# -------------------------------
# Метод скорейшего спуска
# -------------------------------
def steepest_descent_steps(x0, steps=3):
    x = np.array(x0, dtype=float)
    # Матрица Гессе
    A = np.array([[6,2],[2,8]])
    print("\nМетод наискорейшего спуска:")
    for k in range(steps):
        g = grad_f(x)
        alpha = (g @ g) / (g @ (A @ g))
        x = x - alpha * g
        print(f"Шаг {k+1}: x: ({x[0]:.6f}, {x[1]:.6f}), alpha: {alpha:.6f}, градиент: ({g[0]:.6f}, {g[1]:.6f})")
    return x

# Начальная точка
x0 = [2,4]

x_cd = coordinate_descent_steps(x0)
x_gd = gradient_descent_steps(x0)
x_sd = steepest_descent_steps(x0)
```
