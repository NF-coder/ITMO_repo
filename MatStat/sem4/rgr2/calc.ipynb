## 4. Проверка статистических гипотез

### 4.1 Постановка гипотез и схема проверки

### 4.2 Проверка гипотезы о равенстве математических ожиданий

По столбцам X1 и X2 проверим гипотезу о равенстве математических ожиданий двух независимых нормальных совокупностей.

#### Формулировка гипотез

**Нулевая гипотеза (H_0):**
E(X1) = E(X2)

**Альтернативная гипотеза (H_1):**
E(X1) != E(X2)


**Уровень значимости:**
alpha = 0.05

### Выбор критерия

Используем критерий Стьюдента для двух независимых выборок с равными дисперсиями.

**Обоснование:**

1. По условию задачи обе выборки X1 и X2 следуют нормальному распределению
2. Дисперсии совокупностей неизвестны, поэтому используем выборочные оценки
3. Предполагаем, что дисперсии равны (sigma_1^2 = sigma_2^2 = sigma^2)
4. Выборки независимые
5. Объемы выборок n1 = 99, n2 = 99
6. Двусторонняя альтернатива требует двусторонней критической области

**Формула статистики t-критерия:**

t = (mean_X1 - mean_X2) / (S * sqrt(1/n1 + 1/n2))

где:\
S - объединенная (pooled) оценка стандартного отклонения \
При справедливости H_0 статистика t имеет распределение Стьюдента с df = n1 + n2 - 2 степенями свободы.

**Объединённая оценка дисперсии:**

S^2 = ((n1 - 1) * S_1^2 + (n2 - 1) * S_2^2) / (n1 + n2 - 2)

```python
# Читаем данные X1 и X2
X1 = np.array(file.read_column(0, sep=";"))
X2 = np.array(file.read_column(1, sep=";"))

# Основные параметры
n1 = len(X1)
n2 = len(X2)
mean_X1 = np.mean(X1)
mean_X2 = np.mean(X2)
var_X1 = np.var(X1, ddof=1)  # Несмещенная оценка дисперсии S_1^2
var_X2 = np.var(X2, ddof=1)  # Несмещенная оценка дисперсии S_2^2

# Объединённая оценка дисперсии S^2
var_pooled = ((n1 - 1) * var_X1 + (n2 - 1) * var_X2) / (n1 + n2 - 2)
std_pooled = np.sqrt(var_pooled)

# Статистика t-критерия
t_stat = (mean_X1 - mean_X2) / (std_pooled * np.sqrt(1/n1 + 1/n2))

# Степени свободы
df = n1 + n2 - 2

# p-value для двусторонней альтернативы
p_value = 2 * (1 - stats.t.cdf(np.abs(t_stat), df))

# Критическое значение t для alpha 
t_critical = stats.t.ppf(1 - alpha/2, df)
```

## Описательные статистики

```python
print(f"Выборка X1:")
print(f"  Размер: n1 = {n1}")
print(f"  Среднее: mean_X1 = {mean_X1:.4f}")
print(f"  Дисперсия: S_1^2 = {var_X1:.4f}")
print(f"  Стандартное отклонение: S_1 = {np.sqrt(var_X1):.4f}")
print()
print(f"Выборка X2:")
print(f"  Размер: n2 = {n2}")
print(f"  Среднее: mean_X2 = {mean_X2:.4f}")
print(f"  Дисперсия: S_2^2 = {var_X2:.4f}")
print(f"  Стандартное отклонение: S_2 = {np.sqrt(var_X2):.4f}")
```

## Объединённые оценки

```python
print(f"Объединённая дисперсия: S^2 = {var_pooled:.4f}")
print(f"Объединённое стандартное отклонение: S = {std_pooled:.4f}")
print(f"Степени свободы: df = n1 + n2 - 2 = {n1} + {n2} - 2 = {df}")
```

## Статистика критерия

```python
print(f"t_набл = (mean_X1 - mean_X2) / (S * sqrt(1/n1 + 1/n2))")
print(f"t_набл = {t_stat:.4f}")
```

## Критические значения

```python
print(f"alpha = {alpha}")
print(f"Критическое значение: t_cr = t(df={df}, alpha/2={alpha/2}) = {t_critical:.4f}")
print(f"Критическая область (двусторонняя):")
print(f"  W = (-infinity, -{t_critical:.4f}] U [{t_critical:.4f}, +infinity)")
```

## Вывод

```python
if np.abs(t_stat) > t_critical:
    print(f"|t_набл| = {np.abs(t_stat):.4f} > t_cr = {t_critical:.4f}")
    print(f"t_набл находится в критической области W")
    print(f"p-value = {p_value:.6f} < alpha = {alpha}")
    print()
    print("ВЫВОД: Отвергаем H_0")
    print(f"На уровне значимости alpha = {alpha} имеются статистически значимые основания считать, что математические ожидания X1 и X2 различаются.")
else:
    print(f"|t_набл| = {np.abs(t_stat):.4f} <= t_cr = {t_critical:.4f}")
    print(f"t_набл не находится в критической области W")
    print(f"p-value = {p_value:.6f} >= alpha = {alpha}")
    print()
    print("ВЫВОД: Нет оснований отвергнуть H_0")
    print(f"На уровне значимости alpha = {alpha} нет статистически значимых оснований считать, что математические ожидания X1 и X2 различаются.")
```

### 4.4 Непараметрический критерий Манна-Уитни

Для проверки гипотезы о совпадении распределений применим непараметрический критерий Манна-Уитни.

#### Формулировка гипотез

**Нулевая гипотеза (H_0):**
Распределения X1 и X2 совпадают

**Альтернативная гипотеза (H_1):**
Рраспределения X1 и X2 различаются


#### Выбор критерия

Используем непараметрический критерий Манна-Уитни (Mann-Whitney U-test).

```python
# 4.4 - Критерий Манна-Уитни
combined = np.concatenate([X1, X2])
ranks = stats.rankdata(combined, method="average")
ranks_X1 = ranks[:n1]
ranks_X2 = ranks[n1:]

# Суммы рангов
R_X1 = np.sum(ranks_X1)
R_X2 = np.sum(ranks_X2)

# U-статистики
U_X1 = R_X1 - n1 * (n1 + 1) / 2
U_X2 = n1 * n2 - U_X1

mann_whitney_result = stats.mannwhitneyu(X1, X2, alternative="two-sided", method="asymptotic")
U_stat = mann_whitney_result.statistic
p_value_mann_whitney = mann_whitney_result.pvalue

# Критическое значение нормального приближения для двусторонней альтернативы
z_critical = stats.norm.ppf(1 - alpha/2)
z_stat = np.sign(U_stat - n1 * n2 / 2) * stats.norm.ppf(1 - p_value_mann_whitney / 2)
```

## Вывод результатов для критерия Манна-Уитни

```python
print()
print(f"Объёмы выборок: n1 = {n1}, n2 = {n2}")
print(f"Объем объединённой выборки: n1 + n2 = {n1 + n2}")
print()
print(f"Сумма рангов для X1: R_X1 = {R_X1:.1f}")
print(f"Сумма рангов для X2: R_X2 = {R_X2:.1f}")
print()
print(f"U-статистика для X1: U_X1 = {U_X1:.1f}")
print(f"U-статистика для X2: U_X2 = {U_X2:.1f}")
print(f"U_stat = {U_stat:.1f}")
print()
print(f"alpha = {alpha}")
print(f"Критическое значение нормального приближения:")
print(f"  z_cr = {z_critical:.4f}")
print(f"  p-value = {p_value_mann_whitney:.6f}")
print()

# Вывод
if p_value_mann_whitney < alpha:
    print(f"p-value = {p_value_mann_whitney:.6f} < alpha = {alpha}")
    print()
    print("ВЫВОД: Отвергаем H_0")
    print(f"На уровне значимости alpha = {alpha} имеются статистически значимые основания")
    print("считать, что распределения X1 и X2 различаются.")
else:
    print(f"p-value = {p_value_mann_whitney:.6f} >= alpha = {alpha}")
    print()
    print("ВЫВОД: Нет оснований отвергнуть H_0")
    print(f"На уровне значимости alpha = {alpha} нет статистически значимых оснований")
    print("считать, что распределения X1 и X2 различаются.")
```

### 4.3 Проверка гипотезы о дисперсии нормального распределения

По столбцу X3 проверим гипотезу о дисперсии нормального распределения.

#### Формулировка гипотез

**Нулевая гипотеза (H_0):**
sigma_sq = 141.35

**Альтернативная гипотеза (H_1):**
sigma_sq != 141.35

**Уровень значимости:**
alpha = 0.05

#### Выбор критерия

Используем хи-квадрат критерий для проверки гипотезы о дисперсии нормального распределения.

**Обоснование:**

1. По условию задачи выборка X3 следует нормальному распределению
2. Математическое ожидание неизвестно, используем выборочное среднее
3. Двусторонняя альтернатива требует двусторонней критической области

**Формула статистики хи-квадрат критерия:**

chi_sq = (n - 1) * S^2 / sigma_0^2

где:
- S^2 - выборочная дисперсия (несмещённая оценка)
- sigma_0^2 = 141.35 - гипотетическое значение дисперсии
- n - размер выборки

При справедливости H_0 статистика chi_sq имеет распределение хи-квадрат с df = n - 1 степенями свободы.

### Описательные статистики X3

```python
# 4.3 - Проверка гипотезы о дисперсии
sigma_0_sq = 141.35

# Читаем данные X3
X3 = np.array(file.read_column(2, sep=";"))

# Основные параметры
n3 = len(X3)
var_X3 = np.var(X3, ddof=1)  # Несмещённая оценка дисперсии

# Статистика хи-квадрат критерия
chi_sq_stat = (n3 - 1) * var_X3 / sigma_0_sq

# Степени свободы
df3 = n3 - 1

# p-value для двусторонней альтернативы
p_value_3 = 2 * min(stats.chi2.cdf(chi_sq_stat, df3), 1 - stats.chi2.cdf(chi_sq_stat, df3))

# Критические значения для двусторонней альтернативы
chi_sq_critical_left = stats.chi2.ppf(alpha/2, df3)
chi_sq_critical_right = stats.chi2.ppf(1 - alpha/2, df3)
```

### Критические значения для 4.3

```python
print(f"Гипотетическое значение дисперсии: sigma_0^2 = {sigma_0_sq}")
print(f"chi_sq_stat = (n3 - 1) * S_3^2 / sigma_0^2")
print(f"chi_sq_stat = ({n3} - 1) * {var_X3:.4f} / {sigma_0_sq}")
print(f"chi_sq_stat = {chi_sq_stat:.4f}")
```

```python
print(f"alpha = {alpha}")
print(f"Степени свободы: df3 = n3 - 1 = {df3}")
print(f"Критические значения (двусторонняя альтернатива):")
print(f"  chi_sq_left = chi2(df={df3}, alpha/2={alpha/2}) = {chi_sq_critical_left:.4f}")
print(f"  chi_sq_right = chi2(df={df3}, 1-alpha/2={1-alpha/2}) = {chi_sq_critical_right:.4f}")
print(f"Критическая область: W = (0, {chi_sq_critical_left:.4f}] U [{chi_sq_critical_right:.4f}, +infinity)")
```

```python
print(f"chi_sq_stat = {chi_sq_stat:.4f}")
if chi_sq_stat < chi_sq_critical_left or chi_sq_stat > chi_sq_critical_right:
    print(f"chi_sq_stat находится в критической области W")
    print(f"p-value = {p_value_3:.6f} < alpha = {alpha}")
    print()
    print("ВЫВОД: Отвергаем H_0")
    print(f"На уровне значимости alpha = {alpha} имеются статистически значимые основания считать, что дисперсия X3 отличается от {sigma_0_sq}.")
else:
    print(f"chi_sq_stat находится в области допустимых значений")
    print(f"p-value = {p_value_3:.6f} >= alpha = {alpha}")
    print()
    print("ВЫВОД: Нет оснований отвергнуть H_0")
    print(f"На уровне значимости alpha = {alpha} нет статистически значимых оснований считать, что дисперсия X3 отличается от {sigma_0_sq}.")
```

### 4.4 Критерий согласия Пирсона

По столбцу X4 проверим гипотезу о согласии с показательным распределением с параметром lambda = 0.104.

#### Формулировка гипотез

**Нулевая гипотеза (H_0):**
X4 имеет показательное распределение с параметром lambda = 0.104

**Альтернативная гипотеза (H_1):**
X4 не имеет показательного распределения с параметром lambda = 0.104

**Уровень значимости:**
alpha = 0.05

#### Выбор критерия

Используем критерий согласия Пирсона (хи-квадрат критерий).

**Обоснование:**

1. Проверяется согласие выборки с непрерывным законом распределения
2. Критерий Пирсона является универсальным критерием согласия
3. Двусторонняя альтернатива предполагает отклонение в любую сторону

**Формула статистики критерия Пирсона:**

chi_sq = sum((n_i - np_i)^2 / np_i)

где:
- n_i - наблюдаемая частота в i-м интервале
- np_i - ожидаемая частота в i-м интервале при H_0
- Суммирование по всем интервалам

При справедливости H_0 статистика chi_sq имеет распределение хи-квадрат с df = k - s - 1 степенями свободы, где:
- k - число интервалов
- s - число оцениваемых параметров (s = 0, так как lambda задан)

### Описательные статистики X4

```python
# 4.5 - Критерий согласия Пирсона для показательного распределения
lambda_4 = 0.104

# Читаем данные X4
X4 = np.array(file.read_column(3, sep=";"))
n4 = len(X4)

# Разбиваем область значений на 5 интервалов.
# Последний интервал берём до +infinity, потому что показательное распределение задано на [0, +infinity).
num_intervals = 5

base_bounds = []
for i in range(num_intervals):
    left_p = i / num_intervals
    right_p = (i + 1) / num_intervals

    left = -np.log(1 - left_p) / lambda_4 if i != 0 else 0
    right = -np.log(1 - right_p) / lambda_4 if i != num_intervals - 1 else np.inf

    base_bounds.append((left, right))

# Подсчитываем наблюдаемые и ожидаемые частоты
observed_freq = []
expected_freq = []
for left, right in base_bounds:
    if np.isinf(right):
        observed = np.sum(X4 >= left)
        p_interval = np.exp(-lambda_4 * left)
    else:
        observed = np.sum((X4 >= left) & (X4 < right))
        p_interval = np.exp(-lambda_4 * left) - np.exp(-lambda_4 * right)
    observed_freq.append(observed)
    expected_freq.append(p_interval * n4)

observed_freq = np.array(observed_freq)
expected_freq = np.array(expected_freq)

# Объединяем интервалы с малыми ожидаемыми частотами (< 5)
min_expected = 5
observed_combined = []
expected_combined = []
bounds_combined = []
temp_obs = 0
temp_exp = 0
temp_left = None

for i, (left, right) in enumerate(base_bounds):
    if temp_left is None:
        temp_left = left
    temp_obs += observed_freq[i]
    temp_exp += expected_freq[i]
    is_last = (i == len(base_bounds) - 1)

    if temp_exp >= min_expected or is_last:
        if is_last and temp_exp < min_expected and len(observed_combined) > 0:
            observed_combined[-1] += temp_obs
            expected_combined[-1] += temp_exp
            bounds_combined[-1] = (bounds_combined[-1][0], right)
        else:
            observed_combined.append(temp_obs)
            expected_combined.append(temp_exp)
            bounds_combined.append((temp_left, right))
        temp_obs = 0
        temp_exp = 0
        temp_left = None

observed_combined = np.array(observed_combined)
expected_combined = np.array(expected_combined)

# Статистика критерия Пирсона
chi_sq_stat_4 = np.sum((observed_combined - expected_combined)**2 / expected_combined)

# Степени свободы (k - s - 1, где k - число интервалов, s = 0 параметров)
k_intervals = len(observed_combined)
df4 = k_intervals - 1

# p-value
p_value_4 = 1 - stats.chi2.cdf(chi_sq_stat_4, df4)

# Критическое значение
chi_sq_critical_4 = stats.chi2.ppf(1 - alpha, df4)
```

```python
print(f"Выборка X4:")
print(f"  Размер: n4 = {n4}")
print(f"  Среднее: mean_X4 = {np.mean(X4):.4f}")
print(f"  Дисперсия: S_4^2 = {np.var(X4, ddof=1):.4f}")
print(f"  Минимум: min_X4 = {np.min(X4):.4f}")
print(f"  Максимум: max_X4 = {np.max(X4):.4f}")
print(f"\nГипотетический параметр: lambda = {lambda_4}")
```
