```python
import numpy as np
from scipy import stats
```

# Самостоятельная работа №3

Для данных двух независимых выборок проверить гипотезу о равенстве математических ожиданий при предположении, что обе генеральные совокупности нормальны, а их дисперсии неизвестны, но равны. Использовать двусторонний критерий Стьюдента для двух выборок с объединённой оценкой дисперсии. 

```python
def read_csv(path: str) -> list[float]:
    with open(path, "r", encoding="utf-8-sig") as f:
        data = [float(line.split(",")[1]) for line in f.read().split("\n")[1:]]
    return data
```

```python
alpha = 0.05
```

```python
X = read_csv("./variant_8_sample_X.csv")
Y = read_csv("./variant_8_sample_Y.csv")
```

```python
n_x = len(X)
n_y = len(Y)
```

**Гипотезы:** \
H₀: μ_X = μ_Y (математические ожидания равны)\
H₁: μ_X != μ_Y (математические ожидания не равны)

Вычисляем описательные статистики:

```python
mean_x = np.mean(X)
mean_y = np.mean(Y)
var_x = np.var(X, ddof=1)  # несмещённая оценка дисперсии
var_y = np.var(Y, ddof=1)
std_x = np.std(X, ddof=1)
std_y = np.std(Y, ddof=1)

print(f"\n" + "-" * 60)
print("ОПИСАТЕЛЬНЫЕ СТАТИСТИКИ")
print("-" * 60)
print(f"\nВыборка X:")
print(f"  Объём:              n_X = {n_x}")
print(f"  Среднее:            x̄ = {mean_x:.4f}")
print(f"  Дисперсия (смещённая): s²_X = {np.var(X):.4f}")
print(f"  Дисперсия (несмещённая): s²_X = {var_x:.4f}")
print(f"  Стд. отклонение:    s_X = {std_x:.4f}")

print(f"\nВыборка Y:")
print(f"  Объём:              n_Y = {n_y}")
print(f"  Среднее:            ȳ = {mean_y:.4f}")
print(f"  Дисперсия (смещённая): s²_Y = {np.var(Y):.4f}")
print(f"  Дисперсия (несмещённая): s²_Y = {var_y:.4f}")
print(f"  Стд. отклонение:    s_Y = {std_y:.4f}")

# Вычисляем объединённую оценку дисперсии (pooled variance)
s_p_squared = ((n_x - 1) * var_x + (n_y - 1) * var_y) / (n_x + n_y - 2)
s_p = np.sqrt(s_p_squared)

print(f"\n" + "-" * 60)
print("ОБЪЕДИНЁННАЯ ОЦЕНКА ДИСПЕРСИИ")
print("-" * 60)
print(f"\ns²_p = [(n_X - 1)·s²_X + (n_Y - 1)·s²_Y] / (n_X + n_Y - 2)")
print(f"s²_p = [({n_x} - 1)·{var_x:.4f} + ({n_y} - 1)·{var_y:.4f}] / ({n_x} + {n_y} - 2)")
print(f"s²_p = {s_p_squared:.4f}")
print(f"s_p = {s_p:.4f}")

# Вычисляем степени свободы
df = n_x + n_y - 2

# Вычисляем тестовую статистику
t_stat = (mean_x - mean_y) / (s_p * np.sqrt(1/n_x + 1/n_y))

# Вычисляем стандартную ошибку разности средних
se_diff = s_p * np.sqrt(1/n_x + 1/n_y)

print(f"\n" + "-" * 60)
print("ТЕСТОВАЯ СТАТИСТИКА")
print("-" * 60)
print(f"\nСтандартная ошибка разности средних:")
print(f"SE = s_p·√(1/n_X + 1/n_Y) = {s_p:.4f}·√(1/{n_x} + 1/{n_y})")
print(f"SE = {se_diff:.4f}")

print(f"\nТестовая статистика:")
print(f"t = (x̄ - ȳ) / SE")
print(f"t = ({mean_x:.4f} - {mean_y:.4f}) / {se_diff:.4f}")
print(f"t = {t_stat:.4f}")

print(f"\nСтепени свободы: df = n_X + n_Y - 2 = {n_x} + {n_y} - 2 = {df}")

# Вычисляем критическое значение для двусторонного теста
t_critical = stats.t.ppf(1 - alpha/2, df)

# Вычисляем p-value для двусторонного теста
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))

print(f"\n" + "-" * 60)
print("КРИТИЧЕСКИЕ ЗНАЧЕНИЯ И P-VALUE")
print("-" * 60)
print(f"\nДля двустороннего теста при α = {alpha}:")
print(f"Критическое значение: t_кр = ±{t_critical:.4f}")
print(f"p-значение: p = {p_value:.6f}")

print(f"\n" + "-" * 60)
print("ВЫВОД")
print("-" * 60)
print(f"\n|t| = {abs(t_stat):.4f}")
print(f"t_кр = {t_critical:.4f}")
print(f"p-value = {p_value:.6f}")

if abs(t_stat) > t_critical:
    print(f"\nПоскольку |t| > t_кр ({abs(t_stat):.4f} > {t_critical:.4f})")
    print(f"и p-value < α ({p_value:.6f} < {alpha}),")
    print(f"\n✓ ОТКЛОНЯЕМ нулевую гипотезу H₀")
    print(f"\nВЫВОД: При уровне значимости α = {alpha} имеются статистически")
    print(f"значимые отличия между средними значениями выборок X и Y.")
    print(f"Средние математические ожидания НЕ РАВНЫ.")
else:
    print(f"\nПоскольку |t| ≤ t_кр ({abs(t_stat):.4f} ≤ {t_critical:.4f})")
    print(f"и p-value ≥ α ({p_value:.6f} ≥ {alpha}),")
    print(f"\n✗ НЕ ОТКЛОНЯЕМ нулевую гипотезу H₀")
    print(f"\nВЫВОД: При уровне значимости α = {alpha} нет статистически")
    print(f"значимых отличий между средними значениями выборок X и Y.")
    print(f"Нет оснований отклонить предположение о равенстве средних.")

print("\n" + "=" * 60)
```
