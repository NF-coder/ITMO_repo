```python
F = lambda x: (5*x) / (x**4 + 11)
```

```python
l = -2
r = 0
step = 0.2

current = l
sum_x = 0
sum_x2 = 0
sum_y = 0
sum_mul = 0
while True:
    if current > r: break

    print(f"{current:.3f} {F(current):.3f}")
    sum_y += F(current)
    sum_x += current
    sum_x2 += current**2
    sum_mul += F(current)*current
    
    current += step
```

```python
print(f"{sum_x:.3f} {sum_y:.3f} {sum_x2:.3f} {sum_mul:.3f}")
```

```python
l = -2
r = 0
step = 0.2

current = l
sum_x = 0
sum_x2 = 0
sum_x3 = 0
sum_x4 = 0
sum_y = 0
sum_xy = 0
sum_x2y = 0
while True:
    if current > r: break

    print(f"{current:.3f} {F(current):.3f}")
    sum_y += F(current)
    sum_x += current
    sum_x2 += current**2
    sum_x3 += current**3
    sum_x4 += current**4
    sum_xy += F(current)*current
    sum_x2y += F(current)*current**2

    current += step
```

```python
print(f"{sum_x:.3f} {sum_x2:.3f} {sum_x3:.3f} {sum_x4:.3f}\n \
      {sum_y:.3f} {sum_xy:.3f} {sum_x2y:.3f}")
```

```python
from typing_extensions import Callable

def std(
    x: list[float | int],
    y: list[float | int],
    f: Callable[[float | int], float | int]
) -> float | int:
    return sum(map(lambda elem: (f(elem[0]) - elem[1])**2, zip(x,y)))
```

```python
l = -2
r = 0
step = 0.2

x_val_arr = []
y_val_arr = []

current = l
while True:
    if current > r: break

    x_val_arr.append(current)
    y_val_arr.append(F(current))
    
    current += step
```

```python
std(
    x_val_arr, y_val_arr,
    lambda x: 0.205*x - 0.112
)
```

```python
std(
    x_val_arr, y_val_arr,
    lambda x: 0.026 + 0.666 * x + 0.231 * x**2
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-2, 0, 1000)
y_vals = [F(x) for x in x_vals]
f_lin_vals = [0.205*x - 0.112 for x in x_vals]
f_sq_vals = [0.026 + 0.666 * x + 0.231 * x**2 for x in x_vals]


fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(
    x_vals,
    y_vals,
    linestyle='dashed',
    color='orange',
    linewidth=1,
    label='f(x)'
)
ax.plot(
    x_vals,
    f_lin_vals,
    'r-',
    linewidth=2,
    label=r'$a \cdot x + b$',
    alpha=0.8
)
ax.plot(
    x_vals,
    f_sq_vals,
    'g-', 
    linewidth=2,
    label=r'$a_0 + a_1 \cdot x + a_2 \cdot x^2$',
    alpha=0.8
)
ax.scatter(
    x_val_arr,
    y_val_arr,
    color='orange',
    s=40,
    marker='o',
    label='Точки данных'
)


ax.grid(True, alpha=0.3)
ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('f(x)', fontsize=12)
ax.set_title('Функция и её аппроксимации', fontsize=14, fontweight='bold')
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```
